```python
""" Fichier de merge des bases de données pour en obtenir une complète"""

import os
import pandas as pd

df_lic = pd.read_parquet("data_licences/data_licences.parquet")
df_med = pd.read_csv("data_medailles/data_medailles_jo.csv")
```

```python
# Où sont les codes_sport dupliqués ?
dups = df_med[df_med.duplicated("code_sport", keep=False)].sort_values("code_sport")

dups
```

```python
# Créer une table lookup par code_sport
lookup = df_med.groupby("code_sport")["sport"].apply(lambda x: ", ".join(x)).reset_index()

# Merge propre
df = df_lic.merge(lookup, on="code_sport", how="left")

# Si ta colonne sport contient plusieurs disciplines concaténées pour DIV
# ou juste des noms individuels, on remplace par "Divers"
df.loc[df["code_sport"] == "DIV", "sport"] = "Divers"

df.to_parquet("data_complet.parquet")
```

```python
# CAVAPASDUTOUT
df = pd.merge(df_med,df_lic, how='left', on="code_sport")

print(df["sport"].nunique())
df.to_csv("data_complet.csv")
```
