```python
data_licences = pd.read_parquet("data_licences.parquet")
```

```python
data_licences_NR = data_licences[data_licences["code_dep"].isna()]
data_licences_NR.head()
```

```python
data_licences_NR = data_licences[data_licences["departement_long"] == "NR - Non réparti"]
data_licences_NR["code_dep"].describe()
```

```python
ratio_total = data_licences_NR["licences_annuelles"].sum()/data_licences["licences_annuelles"].sum()
ratio_2016 = data_licences_NR["licences_annuelles"][data_licences_NR["annee"] == 2016].sum()/data_licences["licences_annuelles"][data_licences["annee"]==2016].sum()
ratio_2017 = data_licences_NR["licences_annuelles"][data_licences_NR["annee"] == 2017].sum()/data_licences["licences_annuelles"][data_licences["annee"]==2017].sum()
ratio_2018 = data_licences_NR["licences_annuelles"][data_licences_NR["annee"] == 2018].sum()/data_licences["licences_annuelles"][data_licences["annee"]==2018].sum()
ratio_2019 = data_licences_NR["licences_annuelles"][data_licences_NR["annee"] == 2019].sum()/data_licences["licences_annuelles"][data_licences["annee"]==2019].sum()
ratio_2020 = data_licences_NR["licences_annuelles"][data_licences_NR["annee"] == 2020].sum()/data_licences["licences_annuelles"][data_licences["annee"]==2020].sum()
ratio_2021 = data_licences_NR["licences_annuelles"][data_licences_NR["annee"] == 2021].sum()/data_licences["licences_annuelles"][data_licences["annee"]==2021].sum()
ratio_2022 = data_licences_NR["licences_annuelles"][data_licences_NR["annee"] == 2022].sum()/data_licences["licences_annuelles"][data_licences["annee"]==2022].sum()
ratio_2023 = data_licences_NR["licences_annuelles"][data_licences_NR["annee"] == 2023].sum()/data_licences["licences_annuelles"][data_licences["annee"]==2023].sum()
ratio_2024 = data_licences_NR["licences_annuelles"][data_licences_NR["annee"] == 2024].sum()/data_licences["licences_annuelles"][data_licences["annee"]==2024].sum()
print(ratio_total, ratio_2016, ratio_2017, ratio_2018, ratio_2019, ratio_2020, ratio_2021, ratio_2022, ratio_2023, ratio_2024)
```
